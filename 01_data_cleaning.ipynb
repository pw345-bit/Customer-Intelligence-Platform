```python
import pandas as pd

file_path = "data/raw/Online_Retail.xlsx.csv"

df = pd.read_csv(file_path, encoding="latin1")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New dataset shape:", df.shape)
```

```python
df.isnull().sum()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print(df["InvoiceDate"].dtype)
```

```python
df["TotalSales"] = df["Quantity"] * df["UnitPrice"]

df[["Quantity", "UnitPrice", "TotalSales"]].head()
```

```python
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
```

```python
print("Cancelled invoices:",
      df["InvoiceNo"].astype(str).str.startswith("C").sum())
```

```python
df = df[df["UnitPrice"] > 0]

print("Rows after removing invalid prices:", len(df))
```

```python
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

print("Cancelled invoices remaining:",
      df["InvoiceNo"].astype(str).str.startswith("C").sum())
```

```python
df = df[df["Quantity"] > 0]

print("Negative quantities remaining:", (df["Quantity"] < 0).sum())
```

```python
df["TotalSales"] = df["Quantity"] * df["UnitPrice"]
```

```python
df[["Quantity", "UnitPrice", "TotalSales"]].head()
```

```python
print("Final rows:", len(df))
print("Final columns:", len(df.columns))
print("Missing CustomerID:", df["CustomerID"].isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.to_csv(
    "data/processed/online_retail_cleaned.csv",
    index=False
)

print("✅ Cleaned dataset saved successfully!")
```

```python
total_sales = df["TotalSales"].sum()

print(f"Total Sales: £{total_sales:,.2f}")
```

```python
unique_customers = df["CustomerID"].nunique()

print("Unique Customers:", unique_customers)
```

```python
unique_products = df["StockCode"].nunique()

print("Unique Products:", unique_products)
```

```python
print("Number of Countries:", df["Country"].nunique())

df["Country"].value_counts().head(10)
```

```python
top_products = (
    df.groupby("Description")["TotalSales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
df["Month"] = df["InvoiceDate"].dt.to_period("M")

monthly_sales = (
    df.groupby("Month")["TotalSales"]
      .sum()
)

monthly_sales
```

```python
plt.figure(figsize=(12, 6))

monthly_sales.plot()

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
